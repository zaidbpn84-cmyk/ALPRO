```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("Data/voltage.dat")
t = data[:, 0]  # Kolom 0: waktu (s)
V = data[:, 1]  # Kolom 1: tegangan (V)

# Persentase penurunan tegangan
persen_penurunan = ((V[0] - V[-1]) / V[0]) * 100.0

# Estimasi tau dari regresi linear ln(V) = ln(V0) - t / tau
mask = V > 0
slope, intercept = np.polyfit(t[mask], np.log(V[mask]), 1)
tau = -1.0 / slope

# Hitung R^2 untuk memeriksa konsistensi peluruhan eksponensial
ln_V_pred = slope * t[mask] + intercept
ss_res = np.sum((np.log(V[mask]) - ln_V_pred) ** 2)
ss_tot = np.sum((np.log(V[mask]) - np.mean(np.log(V[mask]))) ** 2)
r_squared = 1.0 - (ss_res / ss_tot)

print(f"Tegangan awal         : {V[0]:.4f} V")
print(f"Tegangan akhir        : {V[-1]:.4f} V")
print(f"Persentase penurunan  : {persen_penurunan:.2f}%")
print(f"Perkiraan konstanta τ : {tau:.4f} s")
print(f"Koefisien determinasi : R² = {r_squared:.5f}")

# Plot skala biasa dan skala logaritmik
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5))

ax1.plot(t, V, "o-", color="tab:blue", markersize=4)
ax1.set_xlabel("Waktu, $t$ (s)")
ax1.set_ylabel("Tegangan, $V$ (V)")
ax1.set_title("Skala Linear")
ax1.grid(True, linestyle="--", alpha=0.6)

ax2.plot(t[mask], V[mask], "o", color="tab:orange", markersize=4, label="Data $V(t)$")
ax2.plot(t[mask], np.exp(ln_V_pred), "-", color="black", label=rf"Fit ($\tau = {tau:.3f}\ \mathrm{{s}}$)")
ax2.set_yscale("log")
ax2.set_xlabel("Waktu, $t$ (s)")
ax2.set_ylabel("Tegangan, $V$ (V) [skala log]")
ax2.set_title("Skala Semi-Logaritmik")
ax2.grid(True, which="both", linestyle="--", alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter model kinematika
x0 = 2.0   # posisi awal (m)
v0 = 5.0   # kecepatan awal (m/s)
a = 1.5    # percepatan (m/s^2)

t = np.linspace(0, 10, 80)
x_model = x0 + v0 * t + 0.5 * a * (t ** 2)

# Tambahkan derau acak kecil
rng = np.random.default_rng(42)
derau = rng.normal(loc=0.0, scale=2.0, size=t.shape)
x_data = x_model + derau

# Hitung galat RMS (RMSE)
galat_rms = np.sqrt(np.mean((x_data - x_model) ** 2))
print(f"Galat RMS antara data dan model: {galat_rms:.4f} m")

# Visualisasi
fig, ax = plt.subplots(figsize=(8.5, 5))
ax.scatter(t, x_data, color="tab:orange", alpha=0.8, label="Data Pengukuran Berderau")
ax.plot(t, x_model, color="tab:blue", linewidth=2, label=r"Model $x(t) = x_0 + v_0 t + \frac{1}{2}at^2$")
ax.set_xlabel("Waktu, $t$ (s)")
ax.set_ylabel("Posisi, $x$ (m)")
ax.set_title(f"Perbandingan Model Gerak dan Data Berderau (Galat RMS = {galat_rms:.2f} m)")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


def baca_csv(path):
    try:
        return np.loadtxt(path, delimiter=",", skiprows=1)
    except ValueError:
        return np.loadtxt(path, delimiter=",")


data = baca_csv("electrical.csv")
t = data[:, 0]  # Kolom 0: waktu (s)
V = data[:, 1]  # Kolom 1: tegangan (V)
I = data[:, 2]  # Kolom 2: arus (A)

# Perhitungan daya dan energi
P = V * I
P_max = np.max(P)
P_rata2 = np.mean(P)
energi = np.trapezoid(P, t)

print(f"Daya maksimum : {P_max:.4f} W")
print(f"Daya rata-rata: {P_rata2:.4f} W")
print(f"Total energi  : {energi:.4f} J")

# Plot daya terhadap waktu
fig, ax = plt.subplots(figsize=(8.5, 4.5))
ax.plot(t, P, color="tab:purple", linewidth=1.6, label="Daya Sesaat $P(t) = V(t)I(t)$")
ax.axhline(P_rata2, color="tab:green", linestyle="--", label=f"Daya Rata-rata = {P_rata2:.2f} W")
ax.fill_between(t, P, alpha=0.15, color="tab:purple", label=f"Energi = {energi:.2f} J")
ax.set_xlabel("Waktu, $t$ (s)")
ax.set_ylabel("Daya, $P$ (W)")
ax.set_title("Profil Daya Listrik terhadap Waktu")
ax.grid(True, alpha=0.5)
ax.legend()
plt.tight_layout()
plt.show()

# Simpan hasil ke CSV baru
hasil_gabungan = np.column_stack((t, V, I, P))
np.savetxt(
    "hasil_daya_listrik.csv",
    hasil_gabungan,
    delimiter=",",
    header="waktu_s,tegangan_V,arus_A,daya_W",
    comments="",
    fmt="%.6f",
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter berkas optik Gaussian
I0 = 100.0   # intensitas maksimum (W/m^2)
x0 = 1.5     # pusat berkas (mm)
sigma = 2.0  # simpangan baku spasial (mm)

x = np.linspace(-10, 13, 20001)
I = I0 * np.exp(-((x - x0) ** 2) / (2 * (sigma ** 2)))

# Posisi puncak dan intensitas relatif secara numerik
idx_puncak = np.argmax(I)
x_puncak_num = x[idx_puncak]
I_max_num = I[idx_puncak]
I_relatif = I / I_max_num

# Lebar penuh pada setengah maksimum (FWHM) secara numerik
indeks_setengah_max = np.where(I_relatif >= 0.5)[0]
x_kiri = x[indeks_setengah_max[0]]
x_kanan = x[indeks_setengah_max[-1]]
fwhm_numerik = x_kanan - x_kiri

# FWHM teoritis
fwhm_teoritis = 2.0 * np.sqrt(2.0 * np.log(2.0)) * sigma
galat_fwhm = abs(fwhm_numerik - fwhm_teoritis)

print(f"Posisi puncak numerik : {x_puncak_num:.4f} mm (teoritis: {x0} mm)")
print(f"Intensitas maksimum   : {I_max_num:.4f} W/m²")
print(f"FWHM numerik          : {fwhm_numerik:.6f} mm")
print(f"FWHM teoritis         : {fwhm_teoritis:.6f} mm")
print(f"Selisih absolut FWHM  : {galat_fwhm:.6e} mm")

# Visualisasi
fig, ax = plt.subplots(figsize=(8.5, 4.5))
ax.plot(x, I_relatif, linewidth=2, label="Intensitas Relatif $I(x)/I_{\\max}$")
ax.hlines(0.5, x_kiri, x_kanan, color="red", linewidth=2, label=f"FWHM = {fwhm_numerik:.3f} mm")
ax.axvline(x_puncak_num, color="gray", linestyle="--", label=f"Puncak $x_0 = {x_puncak_num:.2f}$ mm")
ax.set_xlabel("Posisi, $x$ (mm)")
ax.set_ylabel("Intensitas Relatif")
ax.set_title("Profil Intensitas Berkas Gaussian dan Pengukuran FWHM")
ax.grid(True, alpha=0.5)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

# ==============================================================================
# TAHAP 1: PEMBANGKITAN & PEMBACAAN DATA EKSPERIMEN (data -> NumPy)
# ==============================================================================
m = 0.5       # massa (kg)
k = 20.0      # konstanta pegas (N/m)
c = 0.15      # koefisien redaman viskos (N.s/m)
A0 = 0.10     # simpangan awal (m)

gamma = c / (2 * m)
omega0 = np.sqrt(k / m)
omega_teori = np.sqrt(omega0**2 - gamma**2)

t_sim = np.linspace(0, 10, 1000)
rng = np.random.default_rng(7)
x_sim = A0 * np.exp(-gamma * t_sim) * np.cos(omega_teori * t_sim) + rng.normal(0, 0.0015, size=t_sim.shape)

# Simpan dan muat ulang seolah berasal dari berkas sensor mentah
np.savetxt("data_osilasi_mentah.csv", np.column_stack((t_sim, x_sim)), delimiter=",", header="waktu_s,posisi_m", comments="")
data = np.loadtxt("data_osilasi_mentah.csv", delimiter=",", skiprows=1)
t = data[:, 0]
x = data[:, 1]

# ==============================================================================
# TAHAP 2: ANALISIS NUMERIK DENGAN NUMPY (Minimal 3 operasi analisis)
# ==============================================================================
# Operasi 1: Kecepatan sesaat menggunakan gradien numerik
v = np.gradient(x, t)

# Operasi 2: Energi mekanik total E(t) = 1/2 k x^2 + 1/2 m v^2
E_kinetik = 0.5 * m * (v ** 2)
E_potensial = 0.5 * k * (x ** 2)
E_total = E_kinetik + E_potensial

# Operasi 3: Deteksi puncak menggunakan scipy.signal.find_peaks untuk mengatasi noise
# Kita tentukan jarak minimum antar puncak (distance) berdasarkan perkiraan periode teori (~1.0s)
# 1.0s pada t_sim yang memiliki langkah 0.01s berarti sekitar 100 indeks sampel.
puncak_idx, _ = find_peaks(x, distance=80, prominence=0.005)
t_puncak = t[puncak_idx]
periode_rata2 = np.mean(np.diff(t_puncak))
frekuensi_ukur = 1.0 / periode_rata2
omega_ukur = 2.0 * np.pi * frekuensi_ukur

# Operasi 4: Persentase disipasi energi dari awal hingga akhir pengamatan
erengi_hilang_persen = ((E_total[0] - E_total[-1]) / E_total[0]) * 100.0

# ==============================================================================
# TAHAP 3: PEMERIKSAAN KEWAJARAN FISIKA (Satuan, Tanda, Rentang, Referensi)
# ==============================================================================
galat_relatif_omega = abs(omega_ukur - omega_teori) / omega_teori * 100.0

assert np.all(E_total >= 0), "Tidak wajar: Energi mekanik total bernilai negatif."
assert E_total[-1] < E_total[0], "Tidak wajar: Energi mekanik harus meluruh pada sistem teredam."
assert np.max(np.abs(x)) <= A0 * 1.1, "Tidak wajar: Amplitudo melebihi simpangan awal secara signifikan."
assert galat_relatif_omega < 5.0, f"Tidak wajar: Galat frekuensi sudut ({galat_relatif_omega:.2f}%) terlalu besar."

print("=== LAPORAN ANALISIS OSILATOR TEREDAM ===")
print(f"Periode rata-rata terukur  : {periode_rata2:.4f} s")
print(f"Frekuensi sudut terukur    : {omega_ukur:.4f} rad/s (Referensi: {omega_teori:.4f} rad/s)")
print(f"Galat relatif frekuensi    : {galat_relatif_omega:.2f}%")
print(f"Energi mekanik awal        : {E_total[0]:.5f} J")
print(f"Energi mekanik akhir       : {E_total[-1]:.5f} J (Terdisipasi: {erengi_hilang_persen:.2f}%)")
print("Status kewajaran fisika    : LOLOS SEMUA VALIDASI")

# ==============================================================================
# TAHAP 4: VISUALISASI ILMIAH (grafik)
# ==============================================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)

ax1.plot(t, x, color="tab:blue", linewidth=1.2, label="Simpangan $x(t)$")
ax1.plot(t[puncak_idx], x[puncak_idx], "ro", markersize=5, label="Puncak Osilasi Terdeteksi")
ax1.set_ylabel("Simpangan, $x$ (m)")
ax1.set_title("Respons Dinamik dan Peluruhan Energi Osilator Harmonik Teredam")
ax1.grid(True, linestyle="--", alpha=0.5)
ax1.legend()

ax2.plot(t, E_total * 1000, color="tab:green", linewidth=1.5, label="Energi Mekanik Total $E(t)$")
ax2.set_xlabel("Waktu, $t$ (s)")
ax2.set_ylabel("Energi (mJ)")
ax2.grid(True, linestyle="--", alpha=0.5)
ax2.legend()

plt.tight_layout()
plt.savefig("proyek_osilator_teredam.png", dpi=300)
plt.show()

# ==============================================================================
# TAHAP 5: EKSPOR BERKAS HASIL (berkas hasil)
# ==============================================================================
output_data = np.column_stack((t, x, v, E_total))
np.savetxt(
    "hasil_proyek_osilasi.csv",
    output_data,
    delimiter=",",
    header="waktu_s,posisi_m,kecepatan_m_s,energi_mekanik_J",
    comments="",
    fmt="%.6f",
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Memeriksa apakah file ada, jika tidak ada, buat data dummy
file_path = "velocity.csv"
if not os.path.exists(file_path):
    print(f"Peringatan: Berkas '{file_path}' tidak ditemukan. Membuat data simulasi (dummy)...\n")
    # Membuat data dummy untuk simulasi
    t_start = np.linspace(0, 9, 10)
    t_end = t_start + 1.0
    # v = u + a*t dengan sedikit noise (misal a = 1.5, v0 = 2.0)
    t_mid_dummy = 0.5 * (t_start + t_end)
    v_dummy = 2.0 + 1.5 * t_mid_dummy + np.random.normal(0, 0.5, size=t_mid_dummy.shape)

    # Simpan sebagai velocity.csv
    header = "time_start_s,time_end_s,velocity_m_per_s"
    data_to_save = np.column_stack((t_start, t_end, v_dummy))
    np.savetxt(file_path, data_to_save, delimiter=",", header=header, comments="")

# Membaca data dari velocity.csv
data = np.genfromtxt(file_path, delimiter=",", names=True)
t_mid = 0.5 * (data["time_start_s"] + data["time_end_s"])
v = data["velocity_m_per_s"]

slope, intercept = np.polyfit(t_mid, v, 1)
print(f"a  = {slope:.4f} m/s^2")
print(f"v0 = {intercept:.4f} m/s")

fig, ax = plt.subplots(figsize=(6.0, 3.4))
ax.plot(t_mid, v, "s", color="#5A2D6E", label="hasil program C")
ax.plot(t_mid, slope * t_mid + intercept, "-", color="#3F7CAC",
        label=f"garis lurus, $a$ = {slope:.2f} m/s$^2$")
ax.set_xlabel("Waktu titik tengah interval (s)", fontsize=12)
ax.set_ylabel("Kecepatan, $v$ (m/s)", fontsize=12)
ax.tick_params(direction="in", top=True, right=True, labelsize=11)
ax.legend(fontsize=11)
fig.tight_layout()
fig.savefig("kecepatan.pdf")
plt.show()
```
